```python
%sql
SELECT 
    CONCAT(Departure_Station, ' → ', Arrival_Destination) AS Route,
    COUNT(*) AS Services
FROM railway
GROUP BY 
        Departure_Station, 
        Arrival_Destination
ORDER BY Services DESC
LIMIT 5;
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Define consistent color scheme
COLORS = {
    'primary': '#1f77b4',
    'secondary': '#ff7f0e', 
    'tertiary': '#2ca02c',
    'quaternary': '#d62728',
    'purple': '#9467bd',
    'brown': '#8c564b'
}

# Get most popular routes data
df_popular = spark.sql("""
SELECT 
    CONCAT(Departure_Station, ' → ', Arrival_Destination) AS Route,
    COUNT(*) AS Services,
    SUM(Price) AS Total_Revenue
FROM uk_train_ride.train_rise.railway
GROUP BY Departure_Station, Arrival_Destination
ORDER BY Services DESC
LIMIT 5
""").toPandas()

# Create horizontal bar chart
fig, ax = plt.subplots(figsize=(12, 6))
colors = [COLORS['primary'], COLORS['secondary'], COLORS['tertiary'], 
          COLORS['quaternary'], COLORS['purple']]
bars = ax.barh(range(len(df_popular)), df_popular['Services'],
              color=colors, alpha=0.8)
ax.set_yticks(range(len(df_popular)))
ax.set_yticklabels(df_popular['Route'], fontsize=11)
ax.set_title('Top 5 Most Popular Routes', fontsize=16, fontweight='bold', pad=20)
ax.set_xlabel('Number of Services', fontsize=12)
ax.grid(True, alpha=0.3, axis='x')
ax.invert_yaxis()

# Add value labels
for i, (bar, val, rev) in enumerate(zip(bars, df_popular['Services'], df_popular['Total_Revenue'])):
    ax.text(val + 20, bar.get_y() + bar.get_height()/2,
            f'{int(val):,} services | £{int(rev):,}',
            va='center', fontsize=10, fontweight='bold')

plt.tight_layout()
display(plt.show())
```

```python
%sql
SELECT 
    split(Departure_Station, ' ')[0] AS City,
    CONCAT(Departure_Station, ' → ', Arrival_Destination) AS Route,
    
    -- Planned Services (all journeys)
    COUNT(*) AS Planned_Services,
    
    -- Revenue Metrics
    SUM(Price) AS Net_Revenue,
    ROUND(SUM(Price) / COUNT(*), 2) AS Revenue_Per_Trip,
    
    -- Refund Metrics
    SUM(CASE WHEN Refund_Request = 'Yes' THEN Price ELSE 0 END) AS Total_Refunds,
    
    -- On Time Services
    SUM(CASE WHEN Journey_Status = 'On Time' THEN 1 ELSE 0 END) AS On_Time_Services,
    ROUND(
        (SUM(CASE WHEN Journey_Status = 'On Time' THEN 1 ELSE 0 END) * 100.0) / COUNT(*),
        2
    ) AS Reliability_Percentage,
    
    -- Delayed Services
    SUM(CASE WHEN Journey_Status = 'Delayed' THEN 1 ELSE 0 END) AS Delayed_Services,
    ROUND(
        (SUM(CASE WHEN Journey_Status = 'Delayed' THEN 1 ELSE 0 END) * 100.0) / COUNT(*),
        2
    ) AS Delayed_Percentage,
    
    -- Cancelled Services
    SUM(CASE WHEN Journey_Status = 'Cancelled' THEN 1 ELSE 0 END) AS Cancelled_Services,
    ROUND(
        (SUM(CASE WHEN Journey_Status = 'Cancelled' THEN 1 ELSE 0 END) * 100.0) / COUNT(*),
        2
    ) AS Cancelled_Percentage
    
FROM railway
GROUP BY City, Route
ORDER BY City DESC
LIMIT 20;
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Get route reliability data for top routes
df_reliability = spark.sql("""
SELECT 
    CONCAT(Departure_Station, ' → ', Arrival_Destination) AS Route,
    COUNT(*) AS Total_Services,
    ROUND((SUM(CASE WHEN Journey_Status = 'On Time' THEN 1 ELSE 0 END) * 100.0) / COUNT(*), 1) AS On_Time_Pct,
    ROUND((SUM(CASE WHEN Journey_Status = 'Delayed' THEN 1 ELSE 0 END) * 100.0) / COUNT(*), 1) AS Delayed_Pct,
    ROUND((SUM(CASE WHEN Journey_Status = 'Cancelled' THEN 1 ELSE 0 END) * 100.0) / COUNT(*), 1) AS Cancelled_Pct
FROM uk_train_ride.train_rise.railway
GROUP BY Departure_Station, Arrival_Destination
ORDER BY Total_Services DESC
LIMIT 10
""").toPandas()

# Create grouped bar chart
fig, ax = plt.subplots(figsize=(16, 8))
x = range(len(df_reliability))
width = 0.25

bars_ontime = ax.bar([i - width for i in x], df_reliability['On_Time_Pct'],
                     width, label='On Time %', color=COLORS['tertiary'], alpha=0.85)
bars_delayed = ax.bar(x, df_reliability['Delayed_Pct'],
                      width, label='Delayed %', color=COLORS['secondary'], alpha=0.85)
bars_cancelled = ax.bar([i + width for i in x], df_reliability['Cancelled_Pct'],
                        width, label='Cancelled %', color=COLORS['quaternary'], alpha=0.85)

ax.set_title('Route Reliability Comparison (Top 10 Routes by Volume)', fontsize=16, fontweight='bold', pad=20)
ax.set_ylabel('Percentage (%)', fontsize=13)
ax.set_xlabel('Route', fontsize=13)
ax.set_xticks(x)
ax.set_xticklabels([r.split(' → ')[0][:15] + '\n→\n' + r.split(' → ')[1][:15] 
                    for r in df_reliability['Route']], 
                   fontsize=9, ha='center')
ax.legend(fontsize=12, loc='upper right')
ax.grid(True, alpha=0.3, axis='y')
ax.set_ylim(0, 105)

plt.tight_layout()
display(plt.show())
```
